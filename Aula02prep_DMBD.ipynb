```python
import pandas as pd
import matplotlib.pyplot as plt

# Converter a coluna timestamp para datetime
df['timestamp'] = pd.to_datetime(df['timestamp'])

# Organizar os registros cronologicamente
df_ordenado = df.sort_values('timestamp')

# Criar o gráfico
plt.figure(figsize=(15, 8))

plt.plot(
    df_ordenado['timestamp'],
    df_ordenado['corrente_a'],
    color='blue',
    linewidth=1
)

plt.xlabel('Data e hora', fontsize=18)
plt.ylabel('Corrente elétrica (A)', fontsize=18)
plt.title('Comportamento da corrente elétrica ao longo do tempo', fontsize=18)

plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Selecionar somente a máquina M01
df_m01 = df[df['maquina_id'] == 'M01']

# Ordenar por data e hora
df_m01 = df_m01.sort_values('timestamp')

plt.figure(figsize=(15, 8))

plt.plot(
    df_m01['timestamp'],
    df_m01['corrente_a']
)

plt.xlabel('Data e hora')
plt.ylabel('Corrente elétrica (A)')
plt.title('Corrente elétrica da máquina M01')
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import plotly.express as px

# Converter a coluna timestamp para datetime
df['timestamp'] = pd.to_datetime(df['timestamp'])

# Selecionar somente a máquina M01
df_m01 = df[df['maquina_id'] == 'M01']

# Ordenar por data e hora
df_m01 = df_m01.sort_values('timestamp')

# Criar o gráfico
fig = px.line(
    df_m01,
    x='timestamp',
    y='corrente_a',
    title='Corrente elétrica da máquina M01',
    labels={
        'timestamp': 'Data e hora',
        'corrente_a': 'Corrente elétrica (A)'
    }
)

# Configurar a aparência
fig.update_layout(
    width=1200,
    height=600,
    xaxis_title='Data e hora',
    yaxis_title='Corrente elétrica (A)',
    hovermode='x unified'
)

# Exibir gráfico
fig.show()
```

```python
fig = px.line(
    df.sort_values('timestamp'),
    x='timestamp',
    y='corrente_a',
    color='maquina_id',
    title='Corrente elétrica por máquina',
    labels={
        'timestamp': 'Data e hora',
        'corrente_a': 'Corrente elétrica (A)',
        'maquina_id': 'Máquina'
    }
)

fig.update_layout(
    width=1200,
    height=600
)

fig.show()
```
